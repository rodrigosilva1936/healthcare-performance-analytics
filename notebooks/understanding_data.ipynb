```python
import pandas as pd
df = pd.read_csv("C:/Users/rodri/Desktop/Projects/healthcare-performance-analytics/data/raw/healthcare-data.csv", sep=";")
print(df)
```

```python
#Basic Overview 
print("\n SHAPE")
print(df.shape)

print("\n COLUMNS")
print(df.columns)

print("\n DATA TYPES")
print(df.dtypes)

print("\n NULL VALUES")
print(df.isnull().sum())

print("\n DUPLICATED VALUES")
print(df.duplicated().sum())
```

Aqui percebemos que existem 84677 registos e 6 colunas. As colunas são periodo ars, indicador, valor, unidade, id. Não existem valores nulos e também não existem valores duplicados, o que facilita a nossa limpeza no próximo script.

```python
print("\n UNIQUE ARS ")
print(df["ars"].unique())

print("\n NUMBER OF ARS")
print(df["ars"].nunique())

print("\n UNIQUE INDICATORS")
print(df["indicador"].unique())

print("\n NUMBER OF INDICATORS")
print(df["indicador"].nunique())

print("\n UNIQUE UNITS")
print(df["unidade"].unique())
```

Com este overview percebemos que existem 6 Administrações Regionais de Saúde , 4 tipos de indicadores diferentes e 4 unidades distintas também.

```python
# Date Range
df["periodo"] = pd.to_datetime(df["periodo"])
print("\n DATE RANGE")
print("start:", df["periodo"].min())
print("end:", df["periodo"].max())
```

Aqui percebemos que o primeiro registo foi dia 01-11-2016 e o último foi dia 19-09-2026 , ou seja os dados compreendem um intervalo de 10 anos

```python
# Records Per Region
print("\n RECORDS PER ARS")
print(df["ars"].value_counts())

#Records per Indicator
print("\n RECORDS PER INDICATOR")
print(df["indicador"].value_counts())
```

Temos aqui infromações dos group by, e conseguimos perceber que o ARS Algarve tem menos registos. O que pode indicar que existem falhas ao longo dos anos em determinados indicadores.
