```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("C:/Users/rodri/Desktop/Projects/healthcare-performance-analytics/data/processed/healthcare-cleaned.csv",)
```

```python
emergency = df[df["indicador"] == "Número estimado de episódios de urgência"]
emergency.head()
```

emergency fica como um dataframe onde todas as suas linhas correspondem ao numero estimado de ep de urgência.
 

```python
# Number of records by indicator
df["indicador"].value_counts()
```

```python
portugal_emergency = emergency[emergency["ars"] == "Portugal Continental"] 
portugal_emergency.head()
```

```python
yearly_emergency = (portugal_emergency.groupby("ano")["valor"].mean())
yearly_emergency
```

```python
yearly_emergency.plot(title="Average Emergency Episodes by Year - Portugal Continental")
plt.xlabel("Year")
plt.ylabel("Average Number of Emergency Episodes")
plt.show()


```

Este gráfico mostra-nos que houve uma grande queda no número de episódios de urgência em 2020 e uma leve queda do mesmo em 2024-2025. Estes dados só nos dão valores até setembro de 2026, assim sendo, não podemos fazer conclusões precipitadas no ano de 2026 pois o ano ainda não acabou.
