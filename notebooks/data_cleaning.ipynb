```python
import pandas as pd
df = pd.read_csv("C:/Users/rodri/Desktop/Projects/healthcare-performance-analytics/data/raw/healthcare-data.csv", sep=";")
print(df)
```

```python
df["periodo"] = pd.to_datetime(df["periodo"], format= "%Y-%m-%d", errors="raise")
```

Here we can see that the periodo column has all the records in the date format, which will be crucial for cleaning our dataset 
In this data set we saw that there are no duplicated values so using drop_duplicates() is useless, the same goes for the null values and the function dropna()

```python
df["ano"] = df["periodo"].dt.year
df["mes"] = df["periodo"].dt.month
df["trimestre"] = df["periodo"].dt.quarter

```

We need to create the columns ano, mes and trimestre to give more detail and give us more insights especially in powerbi
